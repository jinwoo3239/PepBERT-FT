
# 00. Project Setup

## Project Overview

This project aims to develop and evaluate peptide MIC prediction models using pretrained protein language models and traditional machine learning approaches.

## Project Structure

The project follows a standardized structure to ensure reproducibility, modularity, and consistent result management.

- **Root (`.ipynb`)**: Execute experiments, inspect outputs, and visualize results.
- **`configs/`**: Manage random seeds, hyperparameters, and experiment settings.
- **`src/`**: Implement reusable functions for data processing, tokenization, feature extraction, model training, and evaluation.
- **`data/`**: Store original datasets, processed data, validation splits, and extracted features.
- **`models/`**: Store downloaded pretrained models and fine-tuned model checkpoints separately.
- **`results/`**: Save training histories, predictions, performance metrics, and experiment configurations.
- **`figures/`**: Store figures generated for analysis and publication.

## Experimental Principles

1. All notebooks are located in the project root directory.
2. Python modules (`.py`) contain reusable implementations, while notebooks (`.ipynb`) control execution and result inspection.
3. Random seeds and dataset splits are managed consistently across experiments.
4. Original train/test datasets remain unchanged.
5. Pretrained models are downloaded and stored independently of fine-tuned models.
6. Intermediate outputs, model checkpoints, and final results are saved systematically.
7. All models are evaluated using consistent metrics and experimental protocols.
8. XGBoost is excluded from the machine learning comparison.
